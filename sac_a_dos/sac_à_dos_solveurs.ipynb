# Exemple de résolution d'un problème en nombres entiers

avec https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.milp.

$$\max x_1$$
avec
$$-x_0 + x_1 \leq 1$$
$$3x_0 + 2x_1 \leq 12$$
$$2x_0 + 3x_1 \leq 12$$
$$-3 \leq x_1 \leq 10$$
$$x_0, x_1 \in \mathbb Z$$

```python
c = -np.array([0, 1])
```

```python
A = np.array([[-1, 1], [3, 2], [2, 3]])
b_u = np.array([1, 12, 12])
b_l = np.array([-np.inf,-np.inf,-np.inf])
l = [-np.inf,-3]
u = [np.inf,10]
```

```python
bounds = scopt.Bounds(l,u)
constraints = scopt.LinearConstraint(A, b_l, b_u)
integrality = np.ones_like(c)
```

```python
res = scopt.milp(c=c, constraints=constraints, integrality=integrality, bounds=bounds)
```

```python
#une solution optimale
res.x
```

```python
#valeur de l'optimum
res.fun
```

## sac à dos

Résoudre le problème du sac à dos selon l'exemple ci-dessus avec

| objet  | 1  | 2  | 3  | 4  |
|--------|----|----|----|----|
| valeur | 9  | 11 | 13 | 15 |
| poids  | 6  | 5  | 9  | 7  |

et W = 20

```python
valeurs = np.array([9,11,13,15]) # valeurs des objets
poids = np.array([6,5,9,7]) # poids des objets
W = 20 # capacité du sac à dos
c= -valeurs # on minimise la fonction -valeurs pour maximiser les valeurs
A = np.array([poids]) # contrainte de poids
b_u = np.array([W]) # contrainte supérieure
b_l = np.array([0]) # contrainte inférieure
l = [0,0,0,0] # bornes inférieures (on ne peut pas prendre d'objets négatifs)
u = [1,1,1,1] # bornes supérieures (on peut prendre au plus un exemplaire de chaque objet)
bounds = scopt.Bounds(l,u)
constraints = scopt.LinearConstraint(A, b_l, b_u)
integrality = np.ones_like(c) #toutes les variables sont entières
res = scopt.milp(c=c, constraints=constraints, integrality=integrality, bounds=bounds) # résolution du problème avec milp avec contraintes, bornes et intégralité définies
print("solution optimale", res.x) # affichage de la solution optimale 
print("valeur optimal", -res.fun) # affichage de la valeur optimale (on prend le négatif car on a minimisé -valeurs) 
```

# Relaxation linéaire



Résoudre le même problème sans les contraintes d'intégrité (c'est-à-dire que les variables seront supposées réelles dans [0,1] )

```python
valeurs = np.array([9,11,13,15]) 
poids = np.array([6,5,9,7]) 
W = 20 
c= -valeurs 
A = np.array([poids]) 
b_u = np.array([W]) 
b_l = np.array([0]) 
l = [0,0,0,0] 
u = [1,1,1,1] 
bounds = scopt.Bounds(l,u) 
constraints = scopt.LinearConstraint(A, b_l, b_u) 
res = scopt.milp(c=c, constraints=constraints, bounds=bounds) # résolution du problème sans contrainte d'intégralité
print("solution optimale", res.x) 
print("valeur optimal", -res.fun)
```

```python
def open_sac_à_dos(nomfichier):
    """pour un fichier, renvoie les données du sac à dos"""
    f = open(nomfichier)
    lines = f.readlines()
    val = []
    poids = []
    try:
        l = lines[0].strip().split()
        N, W = int(l[0]), int(l[1])
        for i in range(1,N+1):
            l = lines[i].strip().split()
            val.append(int(l[0]))
            poids.append(int(l[1]))

    except BaseException:
        print("Problème Fichier")
        return
    f.close()
    return N,W,val,poids
```
